```python
# ordinal encoding
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('customer.csv')
df.head()
```

```python
df = df[['review', 'education', 'purchased']]
df.head()
```

```python
df.info()
```

```python
df['review'] = df['review'].astype('category')
df['education'] = df['education'].astype('category')
df['purchased'] = df['purchased'].astype('category')
```

```python
df.info()
```

```python
df['review'].unique()
```

```python
df['education'].unique()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(df[['review', 'education']], df['purchased'], test_size = 0.2, random_state = 42)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories = [['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])
oe.fit(X_train)
X_train = oe.transform(X_train)
X_train
```

```python
X_test = oe.transform(X_test)
X_test
```

```python
oe.categories_
```

```python
# for target variable, always use labelencoder
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
le.fit(Y_train)
Y_train = le.transform(Y_train)
Y_test = le.transform(Y_test)
```
