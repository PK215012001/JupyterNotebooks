```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('cars.csv')
```

```python
df.head()
```

```python
counts = df['brand'].value_counts()
```

```python
rare_cats = counts[counts < 100].index
```

```python
rare_cats
```

```python
df['brand'] = df['brand'].replace(rare_cats, 'others')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.2, random_state= 42)
```

```python
X_train.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop= 'first', sparse_output= False)
```

```python
X_train_trans = ohe.fit_transform(X_train[['brand', 'fuel', 'owner']])
```

```python
X_train_trans
```

```python
ohe.categories_
```

```python
ohe.get_feature_names_out()
```

```python
X_train_trans = pd.DataFrame(X_train_trans, columns= ohe.get_feature_names_out(), index= X_train.index)
```

```python
X_train_trans.head()
```

```python
X_test_trans = ohe.fit_transform(X_test[['brand', 'fuel', 'owner']])
```

```python
X_test_trans = pd.DataFrame(X_test_trans, columns= ohe.get_feature_names_out(), index= X_test.index)
```
