```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('house_train.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
df = df[['GarageQual', 'FireplaceQu', 'SalePrice']]
```

```python
df.head()
```

```python
df.isna().mean()
```

```python
df['GarageQual'].value_counts()
```

```python
df['GarageQual'].value_counts().sort_values(ascending = False).plot.bar()
```

```python
df[df['GarageQual'] == 'TA']['SalePrice'].plot.density(color = 'red', label = 'TA')
df[df['GarageQual'].isna()]['SalePrice'].plot.density(color = 'blue', label = 'NA')
plt.legend()
```

```python
temp = df[df['GarageQual'] == 'TA']['SalePrice']
```

```python
df['GarageQual'].fillna('TA', inplace= True)
```

```python
temp.plot.density(color = 'red', label = 'before')
df[df['GarageQual'] == 'TA']['SalePrice'].plot.density(color = 'blue', label = 'After')
plt.legend()
```

```python
df.head()
```

```python
df['FireplaceQu'].value_counts().sort_values(ascending = False).plot.bar()
```

```python
df[df['FireplaceQu'] == 'Gd']['SalePrice'].plot.density(color = 'red', label = 'Gd')
df[df['FireplaceQu'].isna()]['SalePrice'].plot.density(color = 'blue', label = 'Na')
plt.legend()
```

```python
temp = df[df['FireplaceQu'] == 'Gd']['SalePrice']
df['FireplaceQu'].fillna('Gd', inplace= True)
```

```python
temp.plot.density(color = 'red', label = 'before')
df[df['FireplaceQu'] == 'Gd']['SalePrice'].plot.density(color = 'blue', label = 'after')
plt.legend()
```
