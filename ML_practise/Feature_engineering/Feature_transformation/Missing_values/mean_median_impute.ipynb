```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isna().mean()
```

```python
from sklearn.impute import SimpleImputer
```

```python
si = SimpleImputer()
```

```python
si.fit_transform(df[['Age']])
df
```

```python
df['age_mean'] = si.fit_transform(df[['Age']])
```

```python
df.head()
```

```python
si = SimpleImputer()
```

```python
df['fare_mean'] = si.fit_transform(df[['Fare']])
```

```python
df.head()
```

```python
import seaborn as sns
```

```python
sns.kdeplot(df['Age'], color = 'red', label = 'age')
sns.kdeplot(df['age_mean'], color = 'blue', label = 'age_mean')
plt.legend()
```

```python
sns.kdeplot(df['Fare'], color = 'red', label = 'fare')
sns.kdeplot(df['fare_mean'], color = 'blue', label = 'fare_mean')
plt.legend()
```
