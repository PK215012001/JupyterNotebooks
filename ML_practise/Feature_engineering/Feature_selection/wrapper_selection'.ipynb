```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns= iris.feature_names)
```

```python
df['target'] = iris.target
```

```python
df.head()
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
from mlxtend.feature_selection import ExhaustiveFeatureSelector as EFS
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg = LogisticRegression()

sel = EFS(estimator= lg, max_features= 4, cv = 5, scoring= 'accuracy')
```

```python
sel.fit(X, y)
```

```python
sel.best_score_
```

```python
sel.best_feature_names_
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.2, random_state= 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# creating a base model first
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
std = StandardScaler()
```

```python
X_train = std.fit_transform(X_train)
X_test = std.transform(X_test)
```

```python
lr = LinearRegression()
print(cross_val_score(lr, X_train, y_train, scoring= 'r2', cv= 5, n_jobs= -1).mean())
print(cross_val_score(lr, X_test, y_test, scoring= 'r2', cv= 5, n_jobs= -1).mean())
```

```python
lr = LinearRegression()
efs = EFS(estimator= lr, max_features= 13, print_progress= True, cv= 5, n_jobs = -1, scoring= 'r2')
```

```python
sel = efs.fit(X_train, y_train)
```

```python
sel.best_score_
```

```python
sel.best_feature_names_
```

```python
metric_df = pd.DataFrame.from_dict(sel.get_metric_dict()).T
```

```python
metric_df.head()
```

```python
def adjusted_r2(r2, n_rows, n_features):
    coeff = (n_rows - 1) / (n_rows - n_features -1)
    return 1- (1-r2) * coeff
```

```python
X_train.shape[0]
```

```python
metric_df['num_features'] = metric_df['feature_idx'].apply(lambda x:len(x))
```

```python
metric_df.head()
```

```python
metric_df['adjusted_r2'] = adjusted_r2(metric_df['avg_score'], X_train.shape[0], metric_df['num_features'])
```

```python
metric_df.head()
```

```python
metric_df.sort_values(by= 'adjusted_r2', ascending= False, inplace= True)
```

```python
metric_df.head()
```

```python
X_train_sel = sel.transform(X_train)
X_test_sel = sel.transform(X_test)
```

```python
model = LinearRegression()
print(cross_val_score(model, X_train_sel, y_train, scoring= 'r2', cv = 5).mean())
print(cross_val_score(model, X_test_sel, y_test, cv = 5, scoring = 'r2').mean())
```

```python
# backward elimination
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv')
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.2, random_state= 42)
```

```python
std = StandardScaler()
```

```python
X_train = std.fit_transform(X_train)
X_test = std.transform(X_test)
```

```python
X_train
```

```python
lr = LinearRegression()
print(cross_val_score(lr, X_train, y_train, scoring= 'r2', cv= 5).mean())
print(cross_val_score(lr, X_test, y_test, cv = 5, scoring = 'r2').mean())
```

```python
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
```

```python
sfs = SFS(lr, k_features= 'best', forward= False, floating= False, scoring= 'r2')
```

```python
sfs.fit(X_train, y_train)
```

```python
sfs.k_feature_names_
```

```python
metric_df = pd.DataFrame.from_dict(sfs.get_metric_dict()).T
```

```python
metric_df.head()
```

```python
metric_df['num_features'] = metric_df['feature_idx'].apply(lambda x: len(x)) 
```

```python
metric_df.head()
```

```python
metric_df['adjusted_r2'] = adjusted_r2(metric_df['avg_score'], X_train.shape[0], metric_df['num_features'])
```

```python
metric_df.sort_values(by= 'adjusted_r2', ascending= False, inplace= True)
```

```python
metric_df.head()
```
