```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder, OrdinalEncoder
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.25, random_state= 42, stratify= df.iloc[:, -1])
```

```python
df['review'].value_counts()
```

```python
df['education'].value_counts()
```

```python
df.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
tfs = ColumnTransformer(
                        [('scaling', StandardScaler(), ['age']),
                        ('onehotencoder', OneHotEncoder(drop= 'first', sparse_output= False, handle_unknown= 'ignore'), ['gender']),
                        ('oridnal_encoding', OrdinalEncoder(categories= [['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']]), ['review', 'education']),
                        ]
                        )
```

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression()
```

```python
pipe = Pipeline([
                ('transformation', tfs),
                ('model', lg)
                ])
```

```python
from sklearn.metrics import accuracy_score
```

```python
cvs = cross_val_score(pipe, X_train, y_train, scoring= 'accuracy', cv = 5)
```

```python
print(cvs)
print(cvs.mean())
```

```python
model = pipe.fit(X_train, y_train)
```

```python
model.classes_
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```
