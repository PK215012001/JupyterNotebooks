```python
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, :-1], df.iloc[:, -1], test_size= 0.25, random_state= 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
cvscore = cross_val_score(lr, X_train, y_train, scoring= 'r2', cv= 5)
print(cvscore)
print(cvscore.mean())
```

```python
from sklearn.preprocessing import StandardScaler, Normalizer
```

```python
df.columns
```

```python
ss = StandardScaler()
columns = ['crim', 'indus', 'nox', 'rm', 'age', 'dis', 'tax', 'ptratio', 'b', 'lstat']
X_train[columns] = ss.fit_transform(X_train[columns])
X_test[columns] = ss.transform(X_test[columns])
```

```python
lr2 = LinearRegression()
cvscore2 = cross_val_score(lr2, X_train, y_train, scoring= 'r2', cv= 5)
print(cvscore2)
print(cvscore2.mean())
```

```python
# above results are coming as same becuse of data leakage
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler
```

```python
fs = ColumnTransformer(
                        [('stdscale', StandardScaler(), columns)],
                        remainder= 'passthrough'
                        )
```

```python
pipe = Pipeline(
                [('scaling', fs),
                ('model', lr)]
                )
```

```python
cvscore = cross_val_score(pipe, X_train, y_train, scoring= 'r2', cv= 5)
print(cvscore)
print(cvscore.mean())
```

```python
fs = ColumnTransformer(
                        [('stdscale', MinMaxScaler(), columns)],
                        remainder= 'passthrough'
                        )
```

```python
cvscore = cross_val_score(pipe, X_train, y_train, scoring= 'r2', cv= 5)
print(cvscore)
print(cvscore.mean())
```
