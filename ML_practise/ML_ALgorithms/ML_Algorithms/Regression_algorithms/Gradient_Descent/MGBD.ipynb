```python
import random
class MBGD:

  def __init__(self, batch_size, learning_rate, epochs):
    self.batch_size = batch_size
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.intercept_ = None
    self.coeff_ = None

  def fit(self, X_train, y_train):
    self.intercept_ = 0
    self.coeff_ = np.ones(X_train.shape[1])
    for i in range(self.epochs):
      for j in range(int(X_train.shape[0] / self.batch_size)):
        ids = random.sample(range(X_train.shape[0]), self.batch_size)
        y_hat = self.intercept_ + np.dot(X_train[ids], self.coeff_)

        intercept_der = -2 * np.mean(y_train[ids] - y_hat)
        self.intercept_ = self.intercept_ - (self.learning_rate * intercept_der)

        coeff_der = -2 * np.dot(X_train[ids].T, (y_train[ids] - y_hat))
        self.coeff_ = self.coeff_ - (self.learning_rate * coeff_der)

  def predict(self, X_test):
    return (np.dot(X_test, self.coeff_) + self.intercept_)
```

```python
mbgdregressor = MBGD(X_train.shape[0] // 50, 0.01, 100)
```

```python
mbgdregressor.fit(X_train, y_train)
```

```python
mbgdregressor.intercept_
```

```python
mbgdregressor.coeff_
```

```python
y_pred2 = mbgdregressor.predict(X_test)
```

```python
r2_score(y_test, y_pred2)
```
