```python
# now creating our own sgd regressor class
```

```python
class SGD:

  def __init__(self, learning_rate, epochs):
    self.learning_rate = learning_rate
    self.epochs = epochs
    self.coeff_ = None
    self.intercept_ = None

  def fit(self, X_train, y_train):
    self.intercept_ = 0
    self.coeff_ = np.ones(X_train.shape[1])

    for i in range(self.epochs):
      for j in range(X_train.shape[0]):
        row = np.random.randint(0, X_train.shape[0])
        y_hat = self.intercept_ + np.dot(X_train[row], self.coeff_)

        intercept_der = -2 * (y_train[row] - y_hat)
        self.intercept_ = self.intercept_ - (self.learning_rate * intercept_der)

        coeff_der = -2 * np.dot(X_train[row].T, (y_train[row] - y_hat))
        self.coeff_ = self.coeff_ - (self.learning_rate * coeff_der)

  def predict(self, X_test):
    return np.dot(X_test, self.coeff_) + self.intercept_
```

```python
sgdregressor = SGD(0.01, 50)
```

```python
sgdregressor.fit(X_train, y_train)
```

```python
sgdregressor.intercept_
```

```python
sgdregressor.coeff_
```

```python
y_pred2 = sgdregressor.predict(X_test)
```

```python
r2_score(y_test, y_pred2)
```
