```python
# making a simple gradient descent algorithm
```

```python
import numpy as np
```

```python
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples= 4, n_features= 1, n_informative= 1, n_targets= 1, noise= 80, random_state= 13)
```

```python
X
```

```python
y
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X, y)
```

```python
# now applying linear regression ols
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(X, y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# now we will find the same intercept by using gradient descent by using slope as 78.35
```

```python
# we will start at random value as b = 0
```

```python
b = 0
```

```python
y_pred = ((78.35 * X + b)).reshape(4)
```

```python
plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = '#00808a', label = 'GDS')
plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_slope = -2 *(np.sum(y - m* X.ravel() - b))
```

```python
loss_slope
```

```python
# let's take learning rate as 0.1
lr = 0.1

step_size = lr * loss_slope
step_size
```

```python
# calculating the new intercept
b = b- step_size
```

```python
b
```

```python
y_pred1 = (m * X) + b

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = '#00808a', label = 'itr0')
plt.plot(X, y_pred1, color = '#00fab0', label = 'itr1')
plt.legend()
plt.show()
```

```python
# if we observe it reached near, now again fidning the second iteartion of b

loss_slope = -2 *(np.sum(y - m* X.ravel() - b))
step_size = lr * loss_slope
b = b - step_size
b
```

```python
y_pred2 = (m * X) + b

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = '#00808a', label = 'itr0')
plt.plot(X, y_pred1, color = '#00fab0', label = 'itr1')
plt.plot(X, y_pred2, color = '#00fabc', label = 'itr2')
plt.legend()
plt.show()
```

```python
# if we observe it got even closer
# one more step
loss_slope = -2 *(np.sum(y - m* X.ravel() - b))
step_size = lr * loss_slope
b = b - step_size
b
```

```python
y_pred3 = (m * X) + b

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = '#00808a', label = 'itr0')
plt.plot(X, y_pred1, color = '#00fab0', label = 'itr1')
plt.plot(X, y_pred2, color = '#00fabc', label = 'itr2')
plt.plot(X, y_pred3, color = '#000000', label = 'itr3')
plt.legend()
plt.show()
```

```python
# like these we will reach near to our solution
```

```python
# now doing it in epochs

b = 100
m = 78.35
lr = 0.01
epochs = 200

plt.plot(X, reg.predict(X), color = 'red')

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m * X.ravel() - b)
  b = b - (lr * loss_slope)
  y_pred = m * X + b
  plt.plot(X, y_pred)
```
