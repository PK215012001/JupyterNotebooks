```python
# creating our own class using gradient descent
```

```python
from operator import le
class RidgeGD:

  def __init__(self, alpha, learning_rate, epochs):
    self.alpha = alpha
    self.epochs = epochs
    self.learning_rate = learning_rate
    self.coef_ = None

  def fit(self, X_train, y_train):
    self.intercept_ = 0
    self.coef_ = np.ones(X_train.shape[1]+1).reshape(-1, 1)
    X_train = np.insert(X_train, 0, 1, axis= 1)
    for i in range(self.epochs):
      coeff_der = np.dot(X_train.T, X_train).dot(self.coef_) - np.dot(X_train.T, y_train.reshape(-1, 1)) + self.alpha * self.coef_
      self.coef_ = self.coef_ - (self.learning_rate * coeff_der)

  def predict(self, X_test):
    X_test = np.insert(X_test, 0, 1, axis= 1)
    return (np.dot(X_test, self.coef_.reshape(-1, 1)))
```

```python
ridge = RidgeGD(0.1, 0.001, 100)
```

```python
ridge.fit(X_train, y_train)
```

```python
ridge.coef_
```

```python
r2_score(y_test, ridge.predict(X_test))
```
