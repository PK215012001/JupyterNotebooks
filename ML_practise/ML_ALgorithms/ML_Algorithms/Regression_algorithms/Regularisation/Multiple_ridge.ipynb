```python
import numpy as np
```

```python
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples= 100, n_features= 10, n_targets= 1, n_informative= 10, noise= 20, random_state= 20)
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state= 20, test_size= 0.2)
```

```python
from sklearn.linear_model import Ridge
```

```python
rd1 = Ridge(10)
```

```python
rd1.fit(X_train, y_train)
```

```python
print(rd1.intercept_)
print(rd1.coef_)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test, rd1.predict(X_test))
```

```python
rd2 = Ridge(100)
```

```python
rd2.fit(X_train, y_train)
```

```python
print(rd2.intercept_)
print(rd2.coef_)
```

```python
r2_score(y_test, rd2.predict(X_test))
```

```python
# creating our own class
```

```python
from typing_extensions import Self
class OwnMRidge:

  def __init__(self, alpha):
    self.alpha = alpha
    self.intercept_ = None
    self.coeff_ = None

  def fit(self, X_train, y_train):
    X_train = np.insert(X_train, 0, 1, axis= 1)
    I = np.identity(X_train.shape[1])
    I[0][0] = 0
    # calculating coefficients
    betas = np.linalg.inv(np.dot(X_train.T, X_train) + self.alpha * I).dot(X_train.T).dot(y_train)
    self.intercept_ = betas[0]
    self.coeff_ = betas[1:]

  def predict(self, X_test):
    return (np.dot(X_test, self.coeff_) + self.intercept_)
```

```python
rd1 = OwnMRidge(10)
```

```python
rd1.fit(X_train, y_train)
```

```python
print(rd1.intercept_)
print(rd1.coeff_)
```

```python
r2_score(y_test, rd1.predict(X_test))
```

```python
rd2 = OwnMRidge(100)
```

```python
rd2.fit(X_train, y_train)
```

```python
print(rd2.intercept_)
print(rd2.coeff_)
```

```python
r2_score(y_test, rd2.predict(X_test))
```
