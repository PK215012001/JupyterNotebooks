```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples= 100, n_classes= 2,n_features= 2, n_clusters_per_class= 1, n_informative= 1, n_redundant= 0, n_repeated= 0, random_state= 30, class_sep= 5)
```

```python
X.shape
```

```python
# plotting the data
```

```python
plt.scatter(X[:, 0], X[:, 1], c = y)
```

```python
# doing like perceptron

def step(z):
  return 1 if z >= 0 else 0

def perceptron(x, y):
  x = np.insert(x, 0, 1, axis= 1)
  w = np.ones(x.shape[1])
  lr = 0.01

  for i in range(1000):
    j = np.random.randint(0, x.shape[0])
    y_hat = step(np.dot(w, x[j]))
    w = w + (y[j] - y_hat) * lr * x[j]
  return w[0], w[1:]
```

```python
intercept_,coeff_ = perceptron(X, y)
```

```python
intercept_
```

```python
coeff_
```

```python
m1 = -(coeff_[0]/coeff_[1])
b1 = -(intercept_/coeff_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m1 * x_input + b1
```

```python
plt.scatter(X[:, 0],X[:, 1], c = y)
plt.plot(x_input, y_input)
plt.show()
```

```python
# now doing with sigmoid function ie usign logistic regression
```

```python
def sigmoid(z):
  if (z >= 0):
    return (1/1+np.exp(-z))
  else :
    return np.exp(z) / (1 + np.exp(z))
```

```python
def logistic_reg(x, y):
  x = np.insert(x, 0, 1, axis = 1)
  w = np.ones(x.shape[1])
  lr = 0.01

  for i in range(1000):
    j = np.random.randint(0, x.shape[0])
    y_hat = sigmoid(np.dot(w, x[j]))
    w = w + lr * (y[j] - y_hat) * x[j]
  return w[0], w[1:]
```

```python
int2, coeff2= logistic_reg(X, y)
```

```python
m2 = -(coeff2[0]/coeff2[1])
b2 = -(int2/coeff2[1])
```

```python
x_input_2 = np.linspace(-3, 3, 100)
y_input_2 = m2 * x_input + b2
```

```python
plt.scatter(X[:, 0],X[:, 1], c = y)
plt.plot(x_input, y_input, color = 'red')
plt.plot(x_input_2, y_input_2,color = 'blue')
plt.show()
```
