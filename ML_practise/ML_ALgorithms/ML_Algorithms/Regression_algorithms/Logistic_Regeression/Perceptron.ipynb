```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples= 100, n_features= 2, n_informative= 1, n_redundant= 0, n_repeated=0, n_classes= 2, n_clusters_per_class= 1, random_state= 20,class_sep= 5)
```

```python
X.shape
```

```python
plt.scatter(X[:, 0],X[:, 1], c = y)
plt.show()
```

```python
def step(x):
  return  1 if x >= 0 else 0

def perceptron(X, y):
    X = np.insert(X, 0, 1, axis = 1)
    w = np.ones(X.shape[1])
    lr = 0.01
    for i in range(1000):
      j = np.random.randint(0, X.shape[0])
      y_hat = step(np.dot(w, X[j]))
      w = w + lr * (y[j] - y_hat) * X[j]
    return w[0], w[1:]
```

```python
intercept_,coeff_ = perceptron(X,y)
```

```python
intercept_
```

```python
coeff_
```

```python
m = -(coeff_[0]/coeff_[1])
b = -(intercept_/coeff_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b
```

```python
plt.scatter(X[:, 0],X[:, 1], c = y)
plt.plot(x_input, y_input)
plt.show()
```

```python
# Now doing the same thing by using logistic regression
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg = LogisticRegression()
```

```python
lg.fit(X, y)
```

```python
lg.intercept_
```

```python
lg.coef_
```

```python
m = -(lg.coef_[0][0]/lg.coef_[0][1])
b = -(lg.intercept_/lg.coef_[0][1])
```

```python
X_input = np.linspace(-3, 3, 100)
Y = m*X_input + b
```

```python
plt.scatter(X[:, 0],X[:, 1], c = y)
plt.plot(x_input, y_input,color = 'red')
plt.plot(X_input, Y,color = 'blue')
plt.show()
```
