```python
import numpy as np
import pandas as pd
```

```python
# tosee what's inside the folder
import os
```

```python
os.listdir(path)
```

```python
df = pd.read_csv(path + '/heart.csv')
```

```python
df.head()
```

```python
# aplying both logistic regression and decision tree algortihms
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, y_train, y_test = train_test_split(df.iloc[:, 0:-1], df.iloc[:, -1], random_state= 20, test_size= 0.2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
lg = LogisticRegression()
dt = DecisionTreeClassifier()
```

```python
lg.fit(X_train, y_train)
```

```python
dt.fit(X_train,y_train)
```

```python
lg_pred = lg.predict(X_test)
dt_pred = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, lg_pred)
```

```python
accuracy_score(y_test, dt_pred)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, lg_pred)
```

```python
confusion_matrix(y_test, dt_pred)
```

```python
from sklearn.metrics import  precision_score, recall_score, f1_score
```

```python
print('pr_log:',precision_score(y_test, lg_pred))
print('rec_log:',recall_score(y_test, lg_pred))
print('f1_log:',f1_score(y_test,lg_pred))
```
