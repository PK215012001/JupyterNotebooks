```python
# splitting the data
from sklearn.model_selection import train_test_split
```

```python
X = df['cgpa']
```

```python
y = df['package']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state= 2)
```

```python
# generating a linear model
```

```python
type(X_train)
```

```python
type(X_train.values)
```

```python
from sklearn.linear_model import  LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train.values.reshape(-1, 1), y_train.values.reshape(-1, 1))
```

```python
# now to find metrics, we just havbe to pass the original y and predicted y from our model as shown
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y_test, lr.predict(X_test.values.reshape(-1, 1)))
mae
```

```python
mse = mean_squared_error(y_test, lr.predict(X_test.values.reshape(-1, 1)))
mse
```

```python
r2 = r2_score(y_test, lr.predict(X_test.values.reshape(-1, 1)))
r2
```
