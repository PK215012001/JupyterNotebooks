```python
import numpy as np
import pandas as pd
from plotly import express as px
```

```python
# generating random dataset from make_regression model
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples= 100, n_features= 2, n_informative= 2, random_state= 20, n_targets= 1, noise= 30)
```

```python
X
```

```python
df= pd.DataFrame({'feature1': X[:,0], 'feature2' : X[:,1], 'target':y})
```

```python
df.head()
```

```python
# plotting the data
```

```python
px.scatter_3d(df, x = 'feature1', y= 'feature2', z = 'target')
```

```python
# splitting the data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df[['feature1', 'feature2']], df['target'], test_size = 0.2, random_state= 20)
```

```python
# training the model
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train.values.reshape(-1, 1))
```

```python
# finding the metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mean_absolute_error(y_test, lr.predict(X_test))
```

```python
mean_squared_error(y_test, lr.predict(X_test))
```

```python
r2_score(y_test, lr.predict(X_test))
```
