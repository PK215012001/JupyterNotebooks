```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_samples= 1000, n_features= 3, n_targets= 1, n_informative= 3, noise = 40, random_state= 20)
```

```python
column_names = [f'feature{i}' for i in range(1, 4)]
```

```python
column_names
```

```python
df = pd.DataFrame(X, columns = column_names)
```

```python
df.head()
```

```python
df['target'] = y
```

```python
df.head()
```

```python
fig, ax = plt.subplots(1, 3, figsize = (12, 3))

ax[0].scatter(df['feature1'], df['target'])
ax[1].scatter(df['feature2'], df['target'])
ax[2].scatter(df['feature3'], df['target'])

plt.tight_layout()
plt.show()
```

```python
df[['feature1', 'feature2',  'feature3']].corr()
```

```python
import seaborn as sns
sns.heatmap(df[['feature1', 'feature2',  'feature3']].corr(),annot= True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state= 20)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
errors = y_train - (lr.predict(X_train))
```

```python
sns.kdeplot(errors)
```
