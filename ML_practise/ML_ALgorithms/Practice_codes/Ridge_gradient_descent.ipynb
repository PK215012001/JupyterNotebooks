```python
# now creating our own class
```

```python
from sklearn.base import BaseEstimator, RegressorMixin
class Ridge_descent(BaseEstimator, RegressorMixin):

    def __init__(self, alpha, learning_rate, epochs):
        self.alpha = alpha
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.coeff_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        self.coeff_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        theta = np.insert(self.coeff_, 0, self.intercept_)
        X_train = np.insert(X_train, 0, 1, axis= 1)
        for i in range(self.epochs):
            theta_diff = np.dot(X_train.T, X_train).dot(theta) - np.dot(X_train.T, y_train) + self.alpha * theta
            loss_slope = self.learning_rate * theta_diff
            theta = theta - loss_slope
        self.coeff_ = theta[1:]
        self.intercept_ = theta[0]
        return self

    def predict(self, X_test):
        return np.dot(X_test, self.coeff_) + self.intercept_
```

```python
rd = Ridge_descent(10, 0.001, 10000)
```

```python
rd.fit(X_train, y_train)
```
