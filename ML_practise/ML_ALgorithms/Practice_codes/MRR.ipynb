```python
# Now creating our own class and training the same data, it should give us same results only
```

```python
from sklearn.base import BaseEstimator, RegressorMixin
class MR(BaseEstimator, RegressorMixin):

    def __init__(self, alpha = 1):
        self.alpha = alpha
        self.intercept_ = None
        self.coeff_ = None

    def fit(self, X_train, y_train):
        X_train = np.insert(X_train, 0, 1, 1)
        result = np.linalg.inv(np.dot(X_train.T, X_train) + self.alpha * np.identity(X_train.shape[1])).dot(X_train.T).dot(y_train)
        self.coeff_ = result[1:]
        self.intercept_ = result[0]
        return self

    def predict(self, X_test):
        return np.dot(X_test, self.coeff_) + self.intercept_
```

```python
rd2 = MR(10)
```

```python
rd2.fit(X_train, y_train)
```

```python
rd2.coeff_
```

```python
params = {'alpha' : [0, 0.001, 0.01, 0.1, 1, 10, 100]}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(estimator= MR(), param_grid= params, scoring= 'r2', cv = 5)
```

```python
grid.fit(X_train, y_train)
```
