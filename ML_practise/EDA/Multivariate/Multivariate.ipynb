```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
titanic.head()
```

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
# bivarite analysis between two numerical columns, scatterplots mainly
sns.scatterplot(data = iris, x = 'sepal_length', y = 'sepal_width')
```

```python
# lineplot is better used when we have time column
sns.lineplot(data = flights.groupby('year')['passengers'].sum().reset_index(), x = 'year', y = 'passengers')
```

```python
# we can even do multivariate analysis with hue, style, size parameters
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex', style = 'smoker', size = 'size')
```

```python
titanic.head()
```

```python
# between categorical columns and categorical column
sns.barplot(data = titanic, x = 'Sex', y = 'Age')
```

```python
sns.barplot(data = titanic, x = 'Survived', y = 'Age')
```

```python
sns.barplot(data = titanic, x = 'Survived', y = 'Fare', hue = 'Pclass')
```

```python
sns.boxplot(data = titanic, x = 'Survived', y = 'Age')
```

```python
sns.boxplot(data = titanic, x = 'Survived', y = 'Fare', hue = 'Sex')
```

```python
# Between categorical, categorical, we can do heatmap
sns.heatmap(pd.crosstab(titanic['Survived'], titanic['Sex']))
```

```python
sns.heatmap(pd.crosstab(titanic['Survived'], titanic['Pclass']))
```

```python
sns.clustermap(pd.crosstab(titanic['Survived'], titanic['Pclass']))
```

```python
sns.pairplot(data = iris)
```
