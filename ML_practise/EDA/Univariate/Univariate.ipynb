```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

Categorical columns for categorical columns use count and bar plots

```python
df.info()
```

```python
# coverting categorical cvolumns into categories
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
df['Sex'] = df['Sex'].astype('category')
df['SibSp'] = df['SibSp'].astype('category')
df['Parch'] = df['Parch'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
```

```python
df.info()
```

```python
# vertical bar
df['Survived'].value_counts().plot(kind = 'bar', figsize = (6, 4), title = 'Survival distribution', xlabel = 'Survived', ylabel = 'Count', rot = 0)
```

```python
df['Survived'].value_counts().plot(kind = 'barh')
```

```python
sns.countplot(data = df, x = 'Survived')
```

```python
# Pie chart
df['Survived'].value_counts().plot(kind = 'pie', autopct = '%.2f')
```

```python
# plotting for all columns
df.info()
```

```python
sns.countplot(data = df, y = 'Sex')
```

```python
sns.countplot(data = df, x = 'Embarked')
```

```python
df['Embarked'].value_counts().plot(kind = 'pie', autopct = '%.2f')
```

```python
df['Pclass'].value_counts().plot(kind = 'pie', autopct = '%.2f')
```

```python
# for numerical columns , we can plot histplots or distplots or box plots
df['Age'].plot(kind = 'hist', bins = 10)
```

```python
sns.histplot(df['Age'], bins = 10)
```
