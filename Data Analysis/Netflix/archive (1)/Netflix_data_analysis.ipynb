Summary of the data:
This is a dataset of 8807 rows where each row represnts either a tv show or movie which is being streamed in Netflix. This contains the important data regarding each movie or tv show such as cast, director, duration, title and release date.

Column description:
* show_id : The unique identifier for each show or movie in the tables.
* type : Describes whether the row is either a movie or tv show or any other type.
* title : As name suggests just the title of the respective movie or tv show.
* director : Director or dicrectors of the movie or show.
* cast : Contains cast (actors and actresses mainly and any other important ones if exists) of the movie or show.
* country: Country which produced tha respective content.
* date_added : Content being added in the netflix platform.
* release_year : The official released year of the content in theatres or tv or direct netflix itself.
* rating : Rating in terms of viewer description like adult rated or r rated etc not quality rating like imdb one.
* duration : duration of the content in mins or no.of seasons.
* listed_in : Possibly genre or genres of the content.
* description : A little synopsis about the movie or tv show for teasing purposes.

rating simply means whether it is suitable for teenagers or not and if so what are the conditons(mainly based on sexual themes and language used).
The main categories are:
* G -> General audience (for everyone without any restriction.
* PG -> Parental guidance suggested.
* PG- 13 -> Kids under 13 shouldnot be watched without parental guidance.
* R -> Under 17 requires parents guidance.
* NC-17 -> Strictlu 18+ , no entry for teens.
