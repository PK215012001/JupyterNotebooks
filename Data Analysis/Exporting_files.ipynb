```python
# web scraping , some companies don't provide apis so we will use web scraping and get html file
# scraping ambition box website
url = 'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page=1'
from bs4 import BeautifulSoup

requests.get(url).text
```
