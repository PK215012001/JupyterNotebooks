```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# daily usage of kde in data anlaysis
iris = sns.load_dataset('iris')
iris.head(3)
```

```python
# let's say we want to identify which 2 columns above clearly distinct a certain species
# plotting 4 features kde's 
sns.kdeplot(data = iris, x = 'sepal_length', hue = 'species')
```

```python
sns.kdeplot(data = iris, x = 'sepal_width', hue = 'species')
```

```python
sns.kdeplot(data = iris, x = 'petal_length', hue = 'species')
```

```python
sns.kdeplot(data = iris, x = 'petal_width', hue = 'species')
```

```python
# we can clearly say that petal length and petal width are the 2 features which can clearly distintct.
```

```python
titanic = pd.read_csv('titanic.csv')
titanic.head(3)
```

```python
sns.kdeplot(data = titanic, x = 'Age', hue = 'Survived')
```

```python
# we can also plot cdf plots and use them for confidence of the decision
sns.kdeplot(data = iris, x = 'petal_width', hue = 'species')
sns.ecdfplot(data = iris, x = 'petal_width', hue = 'species')
```

```python
# in 2d plots, the more darker the color, thge more density in that range for both columns
sns.jointplot(data = iris, x = 'sepal_length',  y = 'sepal_width', kind = 'kde', cbar = True, fill = True)
```
