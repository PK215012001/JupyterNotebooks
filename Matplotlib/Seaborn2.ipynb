```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

```python
tips.head(3)
```

```python
# we can even plot scatter plot for numerical vs categorical data
sns.stripplot(data = tips, x = 'day', y = 'total_bill')
```

```python
# fig level
sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill')
```

```python
# adding hue and playing with jitter
sns.catplot(kind = 'strip', data = tips, x = 'day', y = 'total_bill', hue = 'sex', jitter = False)
```

```python
# swarm plots just spread like a swarm
sns.swarmplot(data = tips, x = 'day', y = 'total_bill')
```

```python
sns.swarmplot(data = tips, x = 'day',  y = 'total_bill')
```

```python
# box plots
sns.boxplot(data = tips, y = 'total_bill')
```

```python
# we can even do for other things like below
sns.boxplot(data = tips, x = 'sex', y = 'total_bill')
```

```python
# we can aslo add hue parameter here also
sns.boxplot(data = tips, x = 'sex', y = 'total_bill', hue = 'day')
```

```python
# same can be done by using figure object
sns.catplot(data = tips, kind = 'box', x = 'sex', y = 'total_bill', hue = 'day')
```

```python
# violin plot(box plot + kde plot)
sns.violinplot(data = tips, y = 'total_bill')
```

```python
sns.violinplot(data = tips, x = 'sex', y = 'total_bill')
```

```python
sns.violinplot(data = tips, x = 'sex', y = 'total_bill', hue = 'day')
```

```python
sns.catplot(kind = 'violin', data = tips, x = 'sex', y = 'total_bill', hue = 'day')
```

```python
# we can just make one side only like below
sns.catplot(kind = 'violin', data = tips, x = 'sex', y = 'total_bill', hue = 'day', split = True)
```

```python
sns.catplot(kind = 'violin', data = tips, x = 'day', y = 'total_bill', hue = 'sex', split = True)
```

```python
#bar plot(by default gives us average as aggreghator)
sns.barplot(data = tips,  y ='total_bill')
```

```python
# the small bars are confidence interval bars
sns.barplot(data = tips, x = 'sex', y = 'total_bill')
```

```python
# we can even remove error bars aslo
sns.barplot(data = tips, x = 'sex', y = 'total_bill', errorbar = None)
```

```python
sns.barplot(data = tips, x = 'sex', y = 'total_bill', estimator = 'std')
```

```python
sns.barplot(data = tips, x = 'sex', y = 'total_bill', estimator = 'std', hue = 'smoker')
```

```python
sns.catplot(data = tips, x = 'sex', y = 'total_bill', kind = 'bar', hue = 'smoker', estimator = 'median')
```

```python
# point plot, it is used to see the difference between the aggregate values for the respoective categories
sns.pointplot(data = tips, x = 'sex', y = 'total_bill')
```

```python
sns.catplot(kind = 'point', data = tips, x = 'sex', y = 'total_bill', hue = 'smoker', estimator = 'max')
```

```python
#countplot will just gives no.of observations count for categorical columns
sns.countplot(data = tips, x = 'sex')
```

```python
sns.countplot(data = tips, x = 'sex', hue = 'smoker')
```

```python
sns.catplot(kind = 'count', x = 'sex', hue = 'smoker', data = tips)
```

```python
# here also we can use facetting
sns.catplot(kind = 'count', data =tips, x = 'sex', hue = 'smoker', col = 'time', row = 'day')
```

```python
# regression plot directly gives us the linear regression line
sns.regplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
# figure level function in lmplot and it has hue parameter available
sns.lmplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex')
```

```python
# residual plots(error plots from linear regression)
sns.residplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
# we can also do facetgrid two ways
# normal one
sns.catplot(kind = 'box', data = tips, x = 'sex', y = 'total_bill', hue = 'smoker', col = 'time', row = 'day')
```

```python
# 2nd one is
g = sns.FacetGrid(data = tips, col = 'time', row = 'day')
g.map(sns.boxplot, 'sex', 'total_bill')
```

```python
# pairplot(automatically gives scatter plots between numerical columns)
sns.pairplot(data = iris)
```

```python
sns.pairplot(data = iris, hue = 'species')
```

```python
# same can be done by using pairgrid with extra functionality
g = sns.PairGrid(data = iris, hue = 'species')
g.map(sns.scatterplot)
```

```python
# we can do custyomization for digoanl and off diagonal
g = sns.PairGrid(data  =iris, hue = 'species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g = sns.PairGrid(data = iris, hue = 'species')
g.map_diag(sns.histplot)
g.map_lower(sns.scatterplot)
g.map_upper(sns.boxplot)
```

```python
# we can even mention the numerical columns we want also
g = sns.PairGrid(data = iris, hue = 'species', vars = ['sepal_length', 'sepal_width'])
g.map_diag(sns.histplot)
g.map_lower(sns.scatterplot)
g.map_upper(sns.barplot)
```

```python
# jointplots automatically gives single and double dimensions on two columns
sns.jointplot(data = tips, x = 'total_bill', y = 'tip')
```

```python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip', kind = 'kde', hue = 'sex')
```

```python
# with jointgrid,we can even specify what for combine and what for single plots
g = sns.JointGrid(data = tips, x = 'total_bill', y = 'tip', hue = 'sex')
g.plot(sns.scatterplot, sns.boxplot)
```
