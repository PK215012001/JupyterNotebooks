```python
class User:

    def __init__(self, name, gender):
        self.name = name
        self.gender = gender

    def login(self):
        print(f'Welcome to the website {self.name}')


class Student(User):

    def __init__(self, name, gender, course_name):
        super().__init__(name, gender)
        self.course_name = course_name

    def feedback(self):
        print(f'The  {self.course_name} is actually excellent')


class Teacher(User):

    def __init__(self, name, gender, no_students):
        super().__init__(name, gender)
        self.no_students = no_students

    def reply(self):
        print(f"My name is {self.name} and the no.of students i have under"
              f"my course are: {self.no_students}")
```

```python
s1 = Student('pk', 'male', 'OOP')
s2 = Student('sravani', 'female', 'sql')
```

```python
s1.login()
```

```python
s2.login()
```

```python
s1.feedback()
```

```python
s2.feedback()
```

```python
t1 = Teacher('nitish', 'male', 2)
```

```python
t1.login()
```

```python
t1.reply()
```
