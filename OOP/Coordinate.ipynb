```python
class Point:

  # parametreised constructor
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __str__(self):
    return f"<{self.x}, {self.y}>"

  # distance between two points
  def euclidean_distance(self, other):
    return ((other.y - self.y) **2 + (other.x - self.x) ** 2) ** 0.5

  def distance_orign(self):
    return self.euclidean_distance(Point(0, 0))
```

```python
p2 = Point(3, 4)
```

```python
p1.euclidean_distance(p2)
```

```python
p1.distance_orign()
```

```python
# creating a line class as ax+by+c
class Line():

  def __init__(self, a, b, c):
    self.a = a
    self.b = b
    self.c = c

  def __str__(self):
    return f"{self.a}x + {self.b}y + {self.c} = 0"

  # finding whether a point lies in the line or not
  def point_on_line(self, point):
    if ((self.a * point.x) + (self.b * point.y) + self.c == 0):
      print("point lies on the line")
    else:
      print("Point does not lies on the line")

  def distance_from_point(self, point):
    num = abs((self.a * point.x) + (self.b * point.y) + self.c)
    den = ((self.a **2) + (self.b ** 2)) ** 0.5
    return (num/den)

  def lines_relationship(self, other):
    if ((self.a / other.a) != (self.b / other.b)):
      return "Intersecting lines"
    elif ((self.a / other.a) == (self.b / other.b)):
      if ((self.c / other.c) == (self.a / other.a)):
        return "Same lines"
      else:
        return "Parallel lines"
```
