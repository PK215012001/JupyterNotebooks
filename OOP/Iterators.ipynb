```python
# creating my own range cclass

class my_range:

    def __init__(self, start, end):
        self.start = start
        self.end = end

    # range function is an iterable, so it definately should have an iter metho
    # and it will convert an iterable into iteartor when iter method is called
    # also to run loop we need both start and end which are avialble only in
    # this class, so passing the same object and returning
    def __iter__(self):
        return my_range_iterator(self)


class my_range_iterator:

    def __init__(self, iterable_obj):
        self.iterator = iterable_obj

    # iterator also has an iter method and it again returns an iterator here
    # same class object
    def __iter__(self):
        return self

    # there is also a next function which will do the looping, and for looping 
    # it should have an iterable as an input
    def __next__(self):

        if (self.iterator.start >= self.iterator.end):
            raise StopIteration
        current = self.iterator.start
        self.iterator.start += 1
        return current
```

```python
for i in my_range(1, 10):
    print(i)
```

```python
def gen():
    print('yielding 1')
    yield 1
    print('yielding 2')
    yield 2
    print('yielding 3')
    yield 3
    print('yielding 4')
    yield 4

x = gen()
x
```

```python
print(next(x))
print(next(x))
```

```python
# now the beauty , since we stopped last time at 2,if we run for loop here
# it  will start from 3
for i in x:
    print(i)
```

```python
# same works  for iterator also
x = [1, 2, 3, 4]
it = iter(x)
print(next(it))
print(next(it))
for i in it:
    print(i)
```

```python
# like this we can create our own iterator without using those iter methods
# simply by using yield statment
# own range function by using geneartor
def my_range(start, end):
    current = start
    while (current <= end):
        yield current
        current += 1


for i in my_range(1, 5):
    print(i)
```
