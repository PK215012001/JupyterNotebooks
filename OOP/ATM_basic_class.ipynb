```python
class Atm:

  def __init__(self):
    self.pin = ''
    self.balance = 0
    self.menu()

  def menu(self):
    user_input = input("""
                        1. Press '1' to create pin
                        2. Press '2' to update pin
                        3. Press '3' to view balance
                        4. Press '4' to withdraw balance
                        5. Press 'anything else' to exit
                      """)
    if (user_input == '1'):
      self.create_pin()
    elif (user_input == '2'):
      self.update_pin()
    elif (user_input == '3'):
      self.show_balance()
    elif (user_input == '4'):
      self.withdraw_balance()
    else:
      exit()

  def create_pin(self):
    pin = input('Enter pin: ')
    self.pin = pin
    balance = int(input('Enter balance: '))
    self.balance = balance
    print('Pin created successfully')
    print(f'Current balance is: {self.balance}')
    self.menu()

  def update_pin(self):
    old_pin = input('Enter old pin: ')
    if (old_pin == self.pin):
      new_pin = input('Enter new pin: ')
      self.pin = new_pin
      print('New pin updated successfully')
    else:
      print('Entered pin is wrong')
    self.menu()

  def show_balance(self):
    pin = input('Enter pin: ')
    if (pin == self.pin):
      print(f"Current balance is: {self.balance}")
    else:
      print('Entered pin is wrong')
    self.menu()

  def withdraw_balance(self):
    pin = input('Enter pin: ')
    if (pin == self.pin):
      amount = int(input('Enter amount to withdraw: '))
      if (amount <= self.balance):
        self.balance = self.balance - amount
        print(f'{amount} withdrawn successfully, current balance is: {self.balance}')
      else:
        print(f"Insufficient balance")
    else:
      print('Entered pin is wrong')
    self.menu()
```
