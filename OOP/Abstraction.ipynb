```python
from abc import ABC, abstractmethod

class Bank(ABC):

    def database(self):
        print('Database connected')

    @abstractmethod
    def security(self):
        pass

class mobile_app(Bank):

    def login(self):
        print('mobile app logged')

    def security(self):
        print('secured')
```

```python
m1 = mobile_app()
```

```python
m1.database()
```

```python
m1.login()
```

```python
m1.security()
```
