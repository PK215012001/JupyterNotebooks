```python
# creating a fraction class and using special methods to hijack print, add, sub, mul, divison behaviours
```

```python
class Fraction:

  def __init__(self, num, den):
    self.num = num
    self.den = den

  # hijacks print behaviour
  def __str__(self):
    return f"{self.num}/{self.den}"

  # hijacks '+' behaviour
  def __add__(self, other):
    num = (self.num * other.den) + (other.num * self.den)
    den = self.den * other.den
    return f"{num}/{den}"

  # hijacks '-' behaviour
  def __sub__(self, other):
    num = (self.num * other.den) - (other.num * self.den)
    den = self.den * other.den
    return f"{num}/{den}"

  # hijacks '*' behaviour
  def __mul__(self, other):
    num = self.num * other.num
    den = self.den * other.den
    return f"{num}/{den}"

  # hijacks '/' behaviour
  def __truediv__(self, other):
    num = self.num * other.den
    den = self.den * other.num
    return f"{num}/{den}"
```

```python
fr1 = Fraction(3, 4)
```

```python
print(fr1)
```

```python
fr2 = Fraction(7, 3)
```

```python
print(fr2)
```

```python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
```
