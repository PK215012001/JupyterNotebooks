```python
class ATM:

    __customer_count = 0

    def __init__(self):
        self.__pin = None
        self.__balance = None
        ATM.__customer_count += 1
        self.__customer_id = ATM.__customer_count
        self.menu()

    @staticmethod
    def get_customercount():
        return ATM.__customer_count

    @staticmethod
    def set_customercount(count):
        if (isinstance(count, int) is False | count < 0):
            print('Not possible')
        else:
            ATM.__customer_count = count

    def get_pin(self):
        return self.__pin

    def set_pin(self):
        pin = input('Enter pin: ')
        while (len(pin) != 4):
            print('pin lenght should not  exceed than 4,enter ping again: ')
            pin = input('Enter pin: ')
        else:
            self.__pin = pin

    def get_balance(self):
        return self.__balance

    def set_balance(self):
        balance = int(input('Enter balance: '))
        while (balance < 0):
            print('Invalid balance. Enter balance again: ')
            balance = int(input('Enter balance: '))
        else:
            self.__balance = balance

    def menu(self):
        user_input = input("""
                        1. Press '1' to create pin
                        2. Press '2' to update pin
                        3. Press '3' to view balance
                        4. Press '4' to withdraw balance
                        5. Press 'anything else' to exit
                      """)
        if (user_input == '1'):
            self.create_pin()
        elif (user_input == '2'):
            self.update_pin()
        elif (user_input == '3'):
            self.show_balance()
        elif (user_input == '4'):
            self.withdraw_balance()
        else:
            return

    def create_pin(self):
        if (self.__pin is None and self.__balance is None):
            self.set_pin()
            self.set_balance()
            print('Pin created successfully')
            print(f'Current balance is: {self.__balance}')
            self.menu()
        else:
            print('Bank details already exists')

    def update_pin(self):
        old_pin = input('Enter old pin: ')
        if (old_pin == self.__pin):
            self.set_pin()
            print('New pin updated successfully')
        else:
            print('Entered pin is wrong')
        self.menu()

    def show_balance(self):
        pin = input('Enter pin: ')
        if (pin == self.__pin):
            print(f"Current balance is: {self.__balance}")
        else:
            print('Entered pin is wrong')
        self.menu()

    def withdraw_balance(self):
        pin = input('Enter pin: ')
        if (pin == self.__pin):
            amount = int(input('Enter amount to withdraw: '))
            if (isinstance(amount) != int | amount < 0):
                print('Invalid amount')
            else:
                if (amount <= self.__balance):
                    self.__balance = self.__balance - amount
                    print(f'''{amount} withdrawn successfully,
                            current balance is: {self.__balance}''')
                else:
                    print("Insufficient balance")
        else:
            print('Entered pin is wrong')
        self.menu()
```
